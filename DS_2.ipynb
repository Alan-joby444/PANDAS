use of loc and iloc

```python
data=[1,2,3,4,5,6,7]
s=pd.Series(data)
print(s.loc[1])
```

```python
data=[10,20,30,40,50,60,70]
s=pd.Series(data, index=["A","B","C","D","E","F","G"])
print(s.iloc[2:3])
```

```python
data=[10,20,30,40,50]
s=pd.Series(data, index=["A","B","C","D","E"])
a=s[s>30]
print(a)
print(s.loc["A":"B"])
print(s.iloc[:2])
```

```python
sales=pd.Series([1200,850,1500,950,2000], index=["monday","tuesday","wednesday","thursday","friday"])
print(sales)
```

```python
a=sales.max()
c=sales.min()
print("maximum sales =", a)
print("minimum sales =", c)
```

```python
c=sales[sales<1000]
print("sales less than 1000 =",c)
```

```python
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique())          #shows the unique elements
print(ar.nunique())         #shows the number of unique elements
print(ar.value_counts())    #shows how many times each elements are repeated
```

```python
data=[20,None,30,None,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.isnull()) #shows null values as "true"
print(ar.notnull()) #shows null values as "false"
print(ar.fillna(10)) #fills the  null value with a given number
print(ar.dropna()) #deletes the null values
```
