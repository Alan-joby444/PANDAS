```python
import pandas as pd
data=[20,30,40,50]
s=pd.Series(data)
print(s)
```

the "S" of series should always be in uppercase letter

```python
import pandas as pd
data=[20,30,40,50]
s=pd.Series(data, index=["A","B","C","D"])
print(s)
```

```python
data={"name":"Alan", "age":18}
s=pd.Series(data)
print(s)
```

use of loc and iloc

```python
data=[1,2,3,4,5,6,7]
s=pd.Series(data)
print(s.loc[1])
```

```python
data=[10,20,30,40,50,60,70]
s=pd.Series(data, index=["A","B","C","D","E","F","G"])
print(s.iloc[2:3])
```
