```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("titanic.csv") #to read a file thats imported
df
print(df.head()) #To display the first 5 rows

```

```python
print(df.shape)   #To show the number of rows and columns
```

```python
df.info() #It shows column names,datatypes
```

```python
df.describe() #This gives a statistical summary
```

```python
df.isnull().sum()
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
print(df.isnull().sum())
print("before:",df.shape[0])

df = df.drop_duplicates()

print("after:",df.shape[0])
df = df.rename(columns={"Passengerid":"PassengerID","2urvived":"Survived"})
```

```python
df["Sex"] = df["Sex"].map({0: "Male",1: "Female"})
df
```

```python
df.head()
sns.histplot(data=df, x="Age",bins=20)

plt.title("Age distributon of Titanic passengers")
plt.xlabel("Age")
plt.ylabel("number of passengers")
plt.show()
sns.countplot(data=df, x="Sex", hue= "Survived")

plt.title("Titanic survial by sex")
plt.xlabel("Sex")
plt.ylabel("number of passengers")
plt.show()
sns.countplot(data=df, x="Pclass", hue= "Survived")

plt.title("Titanic survial by passenger class")
plt.xlabel("Passenger class")
plt.ylabel("number of passengers")
plt.show()
```
